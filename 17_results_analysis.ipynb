```python
"""
Скрипт за анализ на влиянието на земното покритие върху тежестта на пожарите.
Показва резултатите в Jupyter и ги запазва като файлове.
"""

import pandas as pd
import numpy as np
from scipy import stats
import os
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from IPython.display import display, HTML

# ====================== НАСТРОЙКИ ======================
%matplotlib inline
plt.rcParams['figure.figsize'] = (12, 8)
sns.set_style("whitegrid")

BASE_DIR = r"D:\data\master_thesis\exports\sentinel2_fire_images\nbr_analysis"
os.chdir(BASE_DIR)
OUTPUT_DIR = os.path.join(BASE_DIR, "analysis_output")
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ====================== 1. ЗАРЕЖДАНЕ ======================
df = pd.read_csv("fire_statistics_summary.csv")

print("=== РАЗМЕРНОСТ И ИНФОРМАЦИЯ ЗА ДАННИТЕ ===")
print(f"Shape: {df.shape}")
df.info()

with open(os.path.join(OUTPUT_DIR, "data_info.txt"), "w", encoding="utf-8") as f:
    f.write(f"Размерност: {df.shape}\n\n")
    import io
    buf = io.StringIO()
    df.info(buf=buf)
    f.write(buf.getvalue())

# ====================== 2. ПОДГОТОВКА НА ДАННИТЕ ======================
покрития = ['Field/Agriculture', 'Coniferous_Forest', 'Deciduous_Forest']
# Кратки имена за таблиците
кратки_имена = {
    'Field/Agriculture': 'Земеделие',
    'Coniferous_Forest': 'Иглолистни',
    'Deciduous_Forest': 'Широколистни'
}
метрики_тежест = ['ModerateSeverity_%', 'HighSeverity_%', 'ModerateHigh_%']

дании_по_метрика = {}
for m in метрики_тежест:
    cols = [f"{p}_{m}" for p in покрития]
    tmp = df[cols].copy()
    tmp.columns = [кратки_имена[p] for p in покрития]
    дании_по_метрика[m] = tmp

# ====================== 3. ДЕСКРИПТИВНА СТАТИСТИКА ======================
print("\n=== ДЕСКРИПТИВНА СТАТИСТИКА ===")
for m in метрики_тежест:
    print(f"\n--- {m} ---")
    display(дании_по_метрика[m].describe().round(2))

with open(os.path.join(OUTPUT_DIR, "descriptive_stats.txt"), "w", encoding="utf-8") as f:
    f.write("Средни стойности на показателите за тежест по тип покритие\n" + "="*60 + "\n")
    for m in метрики_тежест:
        f.write(f"\n--- {m} ---\n")
        f.write(дании_по_метрика[m].describe().round(2).to_string() + "\n")

# ====================== 4. FRIEDMAN TEST ======================
print("\n=== FRIEDMAN TEST ===")
friedman_res = []
for m in метрики_тежест:
    d = дании_по_метрика[m]
    stat, p = stats.friedmanchisquare(d.iloc[:, 0], d.iloc[:, 1], d.iloc[:, 2])
    sig = "ДА" if p < 0.05 else "НЕ"
    friedman_res.append((m, stat, p, sig))
    print(f"{m}: stat={stat:.3f}, p={p:.4f} -> значима разлика? {sig}")

with open(os.path.join(OUTPUT_DIR, "friedman_test.txt"), "w", encoding="utf-8") as f:
    f.write("Friedman test\n" + "="*60 + "\n")
    for m, stat, p, sig in friedman_res:
        f.write(f"{m}: stat={stat:.3f}, p={p:.4f} -> значима разлика? {sig}\n")

# ====================== 5. POST-HOC (WILCOXON С BONFERRONI КОРЕКЦИЯ) ======================
alpha_bonf = 0.05 / 3
print("\n=== POST-HOC (WILCOXON С BONFERRONI КОРЕКЦИЯ) ===")
for m in метрики_тежест:
    d = дании_по_метрика[m]
    _, p_f = stats.friedmanchisquare(d.iloc[:, 0], d.iloc[:, 1], d.iloc[:, 2])
    if p_f >= 0.05:
        print(f"{m}: Friedman незначим – пропускаме post-hoc.")
        continue
    print(f"\n--- {m} (Friedman p={p_f:.4f}) ---")
    pairs = [(0, 1), (0, 2), (1, 2)]
    names = d.columns  # вече ще са 'Земеделие', 'Иглолистни', 'Широколистни'
    for i, j in pairs:
        w_stat, w_p = stats.wilcoxon(d.iloc[:, i], d.iloc[:, j])
        sig = "ДА" if w_p < alpha_bonf else "НЕ"
        print(f"{names[i]} vs {names[j]}: W={w_stat:.3f}, p={w_p:.4f} -> значима? {sig}")

with open(os.path.join(OUTPUT_DIR, "posthoc_wilcoxon.txt"), "w", encoding="utf-8") as f:
    f.write("Post-hoc двойни сравнения (Wilcoxon, Bonferroni)\n" + "="*60 + "\n")
    for m in метрики_тежест:
        d = дании_по_метрика[m]
        _, p_f = stats.friedmanchisquare(d.iloc[:, 0], d.iloc[:, 1], d.iloc[:, 2])
        if p_f >= 0.05:
            f.write(f"\n{m}: Friedman незначим.\n")
            continue
        f.write(f"\n--- {m} (Friedman p={p_f:.4f}) ---\n")
        pairs = [(0, 1), (0, 2), (1, 2)]
        names = d.columns
        for i, j in pairs:
            w_stat, w_p = stats.wilcoxon(d.iloc[:, i], d.iloc[:, j])
            sig = "Да" if w_p < alpha_bonf else "Не"
            f.write(f"{names[i]} vs {names[j]}: W={w_stat:.3f}, p={w_p:.4f} -> значима? {sig}\n")

# ====================== 6. КОРЕЛАЦИЯ МЕЖДУ %_of_Total И dNBR_Mean ======================
print("\n=== КОРЕЛАЦИЯ МЕЖДУ %_of_Total И dNBR_Mean ===")
for p in покрития:
    col = f"{p}_%_of_Total"
    r, pv = stats.pearsonr(df[col], df['dNBR_Mean'])
    print(f"{кратки_имена[p]}: r={r:.3f}, p={pv:.4f}")

with open(os.path.join(OUTPUT_DIR, "correlations.txt"), "w", encoding="utf-8") as f:
    f.write("Корелация (Pearson) между %_of_Total и dNBR_Mean\n" + "="*60 + "\n")
    for p in покрития:
        col = f"{p}_%_of_Total"
        r, pv = stats.pearsonr(df[col], df['dNBR_Mean'])
        f.write(f"{кратки_имена[p]}: r={r:.3f}, p={pv:.4f}\n")

# ====================== 7. ВИЗУАЛИЗАЦИИ ======================
# 7.1 Boxplots
fig, axes = plt.subplots(1, 3, figsize=(16, 6))
for ax, m in zip(axes, метрики_тежест):
    дании_по_метрика[m].boxplot(ax=ax)
    ax.set_title(f"Разпределение на {m}")
    ax.set_ylabel("% от пикселите на покритието")
    ax.tick_params(axis='x', rotation=0)
plt.suptitle("Boxplots на тежестта по тип земно покритие", fontsize=16)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "boxplots_severity.png"), dpi=150)
plt.show()

# 7.2 Scatter plots
fig, axes = plt.subplots(3, 2, figsize=(14, 12))
for i, p in enumerate(покрития):
    vul_col = f"{p}_Vulnerability_Index"
    pct_col = f"{p}_%_of_Total"
    sns.regplot(x=df['dNBR_Mean'], y=df[vul_col], ax=axes[i, 0], ci=95)
    axes[i, 0].set_title(f"{кратки_имена[p]}: Vulnerability vs dNBR_Mean")
    sns.regplot(x=df[pct_col], y=df[vul_col], ax=axes[i, 1], ci=95)
    axes[i, 1].set_title(f"{кратки_имена[p]}: Vulnerability vs %_of_Total")
plt.suptitle("Зависимост на индекса на уязвимост от dNBR и % от площта", fontsize=16)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "scatter_vulnerability.png"), dpi=150)
plt.show()

# 7.3 Корелационна матрица
corr_cols = ['dNBR_Mean', 'Total_Burned_%'] + [f"{p}_%_of_Total" for p in покрития] + [f"{p}_Vulnerability_Index" for p in покрития]
corr_df = df[corr_cols].rename(columns=lambda c: c.replace('/','_').replace(' ','_'))
plt.figure(figsize=(10, 8))
sns.heatmap(corr_df.corr(), annot=True, cmap='RdBu_r', center=0, fmt=".2f")
plt.title("Корелационна матрица на основни променливи")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "correlation_heatmap.png"), dpi=150)
plt.show()

# ====================== 8. ЛИНЕЙНА РЕГРЕСИЯ ======================
def calculate_pvalues(X, y, coefficients):
    n, k = X.shape
    y_pred = X @ coefficients
    residuals = y - y_pred
    df_resid = n - k
    sigma2 = np.sum(residuals**2) / df_resid
    cov_matrix = sigma2 * np.linalg.inv(X.T @ X)
    std_errors = np.sqrt(np.diag(cov_matrix))
    t_stats = coefficients / std_errors
    p_values = 2 * (1 - stats.t.cdf(np.abs(t_stats), df=df_resid))
    return p_values, std_errors

def calculate_vif(X):
    vif = []
    n_cols = X.shape[1]
    for i in range(n_cols):
        y_i = X[:, i]
        X_i = np.delete(X, i, axis=1)
        reg = LinearRegression().fit(X_i, y_i)
        r2 = reg.score(X_i, y_i)
        vif.append(1 / (1 - r2))
    return np.array(vif)

print("\n=== ЛИНЕЙНА РЕГРЕСИЯ ЗА ПРЕДСКАЗВАНЕ НА VULNERABILITY INDEX ===")
reg_summary = []

for p in покрития:
    print(f"\n--- Модел за {кратки_имена[p]} ---")
    target = f"{p}_Vulnerability_Index"
    pct_col = f"{p}_%_of_Total"

    X = df[['dNBR_Mean', 'Total_Burned_%', pct_col]].values
    y = df[target].values
    X_design = np.column_stack([np.ones(X.shape[0]), X])
    var_names = ['const', 'dNBR_Mean', 'Total_Burned_%', f"{кратки_имена[p]}_%_of_Total"]

    model = LinearRegression().fit(X_design, y)
    coeffs = model.coef_
    y_pred = model.predict(X_design)
    r2 = r2_score(y, y_pred)
    adj_r2 = 1 - (1 - r2) * (len(y) - 1) / (len(y) - X_design.shape[1] - 1)

    pvals, std_errs = calculate_pvalues(X_design, y, coeffs)
    vif_values = calculate_vif(X_design[:, 1:])

    print(f"R² = {r2:.4f},  Adj. R² = {adj_r2:.4f}")
    coef_df = pd.DataFrame({
        'Коеф.': coeffs,
        'Стд. грешка': std_errs,
        't-статистика': coeffs / std_errs,
        'p-стойност': pvals
    }, index=var_names)
    display(coef_df.round(4))
    print("VIF (без константа):")
    vif_df = pd.DataFrame({'Променлива': var_names[1:], 'VIF': vif_values})
    display(vif_df.round(2))

    # Диагностични графики
    residuals = y - y_pred
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
    stats.probplot(residuals, dist="norm", plot=ax1)
    ax1.set_title(f"QQ-plot - {кратки_имена[p]}")
    ax2.scatter(y_pred, residuals)
    ax2.axhline(0, color='red', linestyle='--')
    ax2.set_xlabel("Предсказани стойности")
    ax2.set_ylabel("Остатъци")
    ax2.set_title(f"Остатъци vs предсказани - {кратки_имена[p]}")
    plt.suptitle(f"Диагностика на регресионен модел за {кратки_имена[p]}", fontsize=14)
    plt.tight_layout()
    plt.savefig(os.path.join(OUTPUT_DIR, f"diagnostics_{p.replace('/','_')}.png"), dpi=150)
    plt.show()

    reg_summary.append(f"{кратки_имена[p]}: R² = {r2:.3f}, Adj.R² = {adj_r2:.3f}")

    out_path = os.path.join(OUTPUT_DIR, f"regression_{p.replace('/','_')}.txt")
    with open(out_path, "w", encoding="utf-8") as f:
        f.write(f"Линейна регресия за {кратки_имена[p]}\n" + "="*60 + "\n")
        f.write(f"R² = {r2:.4f},  Adj. R² = {adj_r2:.4f}\n\n")
        f.write("Коефициенти:\n")
        f.write(f"{'Променлива':<25} {'Коеф.':>10} {'Стд. грешка':>12} {'t-стат.':>10} {'p-стойност':>12}\n")
        for name, coef, se, pv in zip(var_names, coeffs, std_errs, pvals):
            f.write(f"{name:<25} {coef:>10.4f} {se:>12.4f} {coef/se:>10.4f} {pv:>12.4f}\n")
        f.write("\nVIF (без константа):\n")
        for vname, vif in zip(var_names[1:], vif_values):
            f.write(f"{vname:<25} {vif:>10.2f}\n")

# Обобщение
print("\n=== ОБОБЩЕНИЕ НА РЕГРЕСИОННИТЕ МОДЕЛИ ===")
for line in reg_summary:
    print(line)

with open(os.path.join(OUTPUT_DIR, "regression_summary.txt"), "w", encoding="utf-8") as f:
    f.write("Кратко обобщение на регресионните модели\n" + "="*60 + "\n")
    for line in reg_summary:
        f.write(line + "\n")

print(f"\n✅ Всички резултати са записани в папка: {OUTPUT_DIR}")
```
